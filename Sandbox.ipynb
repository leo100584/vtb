# Sandbox for Vega Altair med bruk av hjemmelaget data

```python
import pandas as pd
import altair as alt

from IPython.display import display

# Eksempeldata
df = pd.DataFrame({
    "Areal": [55, 65, 72, 80, 95, 110, 120, 135, 150, 175],
    "Pris": [3.2, 3.8, 4.1, 4.5, 5.3, 6.2, 6.8, 7.5, 8.4, 9.8],
    "Boligtype": [
        "Leilighet", "Leilighet", "Leilighet",
        "Rekkehus", "Rekkehus",
        "Enebolig", "Enebolig",
        "Enebolig", "Enebolig", "Enebolig"
    ],
    "By": [
        "Oslo", "Oslo", "Bergen",
        "Bergen", "Trondheim",
        "Oslo", "Trondheim",
        "Bergen", "Oslo", "Trondheim"
    ]
})

# Markeringsverktøy
brush = alt.selection_interval()

# Scatterplot
scatter = (
    alt.Chart(df)
    .mark_circle(size=600)
    .encode(
        x=alt.X('Areal:Q', title='Areal (m²)'),
        y=alt.Y('Pris:Q', title='Pris (mill. kr)'),
        color=alt.condition(
            brush,
            'Boligtype:N',
            alt.value('lightgray')
        ),
        tooltip=[
            'Areal',
            'Pris',
            'Boligtype',
            'By'
        ]
    )
    .add_params(brush)
    .properties(
        width=800,
        height=450,
        title='Boligpriser vs. areal'
    )
)

# Gjennomsnittspris for markerte boliger
stats = (
    alt.Chart(df)
    .mark_bar()
    .encode(
        x=alt.X(
            'mean(Pris):Q',
            title='Gjennomsnittlig pris (mill. kr)'
        )
    )
    .transform_filter(brush)
    .properties(
        width=600,
        height=80,
        title='Snittpris for markerte boliger'
    )
)

chart = scatter & stats
display(chart)
```
